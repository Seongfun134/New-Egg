```python
# Import chrome webdriver to access chrome options
# Import chrome webdriver to access chrome options
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
import pandas as pd
import time
import selenium
from selenium import webdriver
```

```python
url= 'https://www.newegg.com/GPUs-Video-Graphics-Cards/SubCategory/ID-48'
service = Service(ChromeDriverManager().install())
driver = webdriver.Chrome(service=service)
driver.get(url)
time.sleep(3)
cards= driver.find_elements(By.CSS_SELECTOR , 'div.item-cell')
data =[]
#Loop through each card and extract fields
for card in cards:
##Field 1
    try:
        image_url = card.find_element(By.CSS_SELECTOR , 'a.item-img img').get_attribute("src")
    except:
        image_url = None 
    ##Field 2 Rating 
    try:
        rating = card.find_element(By.CSS_SELECTOR , 'a.item-rating').get_attribute("title")
    except:
        rating= None 
    try :
        review = card.find_element(By.CSS_SELECTOR , 'span.item-rating-num').text.strip("()")
    except: 
        review = None
    #Field 3 : Manufacturer (title attribute, fallback alt)
    try:
        brand_img=card.find_element(By.CSS_SELECTOR , 'a.item-brand img')
        brand = brand_img.get_attribute('title') or brand_img.get_attribute("alt")
    except:
        brand = None
    #Field 4  Product Title 
    try:
        title= card.find_element(By.CSS_SELECTOR , 'a.item-title').text
    except:
        title = None
    ## Field 5: Model number (split text after "Model #:")
    try:
        features_text = card.find_element(By.CSS_SELECTOR ,'ul.item-features').text
        model =features_text.split("Model #:")[-1].strip() if "Model #:" in features_text else None 
    except:
        model = None 
    ## Field 6 Current Price (dollars + cents combined)
    try:
        price_el = card.find_element(By.CSS_SELECTOR , 'li.price-current')
        dollars =price_el.find_element(By.TAG_NAME , "strong")
        cents=price_el.find_element(By.TAG_NAME , "sup")
        price =f"{dollars.text}{cents.text}"
    except:
        price = None
 ## Field 7: shipping info
    try:
        shipping = card.find_element(By.CSS_SELECTOR,'li.price-ship').text
    except:
        shipping = None

    ## Save this card's data (runs for every card, not just failures)
    data.append({
        "Title": title, "Brand": brand, "Rating": rating,
        "Review_Count": review, "Model": model, "Price": price,
        "Shipping": shipping, "Image_URL": image_url
    })

df = pd.DataFrame(data)
print(df)
df.to_csv("newegg_products.csv", index=False)
driver.quit()
```
